**EJEMPLO 1**

Una institución educativa recibe mensajes de alumnos que necesitan ayuda con sus equipos de cómputo. El agente inteligente debe clasificar cada mensaje en una categoría.

```python

#def: permite definir una función.
      #mensaje: es el dato que recibe la función.
def clasificar_solicitud(mensaje):
            #lower: convierte el texto a minúsculas.
    mensaje = mensaje.lower()

    #if y elif: permiten evaluar condiciones.
    if "no enciende" in mensaje or "pantalla" in mensaje:
        # return: devuelve la categoría identificada.
        return "Hardware"

    elif "internet" in mensaje or "wifi" in mensaje:
        return "Conectividad"

    elif "contraseña" in mensaje or "contrasena" in mensaje:
        return "Acceso a cuenta"
    #else: establece una categoría cuando no se cumple ninguna condición anterior.
    else:

        return "Solicitud no identificada"

#Imprime el resultado y manda el mensaje a la función para
#obtener una respuesta
print(clasificar_solicitud("Mi computadora no enciende"))
print(clasificar_solicitud("No tengo internet"))
print(clasificar_solicitud("Olvidé mi contraseña"))
print(clasificar_solicitud("Necesito instalar una impresora"))
```

**EJEMPLO 2**

El agente inteligente de soporte debe determinar qué incidentes requieren atención inmediata.

```python
def clasificar_prioridad(incidente):
    incidente = incidente.lower()

    if "servidor caído" in incidente or "servidor caido" in incidente:
        return "Alta"

    elif "no puedo ingresar" in incidente:
        return "Media"

    elif "cambiar color" in incidente:
        return "Baja"

    else:
        return "Por evaluar"
```

```python
#ENVIO DE INCIDENTES

incidentes = [
    "El servidor está caído",
    "No puedo ingresar a la plataforma",
    "Quiero cambiar color de mi perfil",
    "La impresora hace ruido"
]

for incidente in incidentes:
    prioridad = clasificar_prioridad(incidente)
    print(incidente, "->", prioridad)
```

**EJERCICIO 3**

Desarrollar un agente inteligente que analice las solicitudes de los usuarios de una biblioteca escolar, las clasifique según el servicio requerido y determine qué acción debe realizar.

Una biblioteca escolar recibe mensajes de estudiantes que necesitan pedir libros prestados, devolver materiales, renovar préstamos o reportar libros extraviados. El personal desea automatizar la clasificación inicial de estas solicitudes.

Crea una función clasificadora en Python que permita al agente identificar el tipo de solicitud y recomendar una acción.

**REGLAS DE CLASIFICACIÓN**

| Categoría | Palabras clave | Acción del agente |
|---|---|---|
| Préstamo | prestar, préstamo, llevar libro | Registrar solicitud de préstamo |
| Devolución | devolver, entregar libro | Registrar devolución |
| Renovación | renovar, ampliar préstamo | Revisar posibilidad de renovación |
| Extravío | perdido, extravié, extraviado | Enviar a revisión del bibliotecario |
| Consulta | horario, ubicación, catálogo | Proporcionar información |
| No identificada | Ninguna coincidencia | Solicitar más información |


El agente deberá:
1. Solicitar al usuario que escriba un mensaje.
2. Convertir el mensaje a minúsculas.
3. Utilizar una función clasificadora para identificar la categoría.
4. Utilizar otra función para determinar la acción recomendada.
5. Mostrar un reporte con el mensaje, la categoría y la acción.

```python
#ESCRIBE TU CÓDIGO AQUÍ
```

```python
import json

def clasificar_solicitud(mensaje):
    texto = mensaje.lower()

    if any(p in texto for p in ["extravié", "extraviado", "perdido", "perdí"]):
        return "Extravío"

    elif any(p in texto for p in ["renovar", "ampliar préstamo", "ampliar el préstamo"]):
        return "Renovación"

    elif any(p in texto for p in ["devolver", "entregar libro", "entregar el libro"]):
        return "Devolución"

    elif any(p in texto for p in ["préstamo", "prestamo", "prestar", "llevar libro"]):
        return "Préstamo"

    elif any(p in texto for p in ["horario", "ubicación", "ubicacion", "catálogo", "catalogo"]):
        return "Consulta"

    else:
        return "Solicitud no identificada"


def generar_correo_soporte(mensaje, categoria):
    destinatario = "soporte.biblioteca@institucion.edu"

    if categoria == "Extravío":
        prioridad = "Alta"
        accion = "Revisar el reporte de material extraviado."
    elif categoria == "Renovación":
        prioridad = "Media"
        accion = "Verificar si el préstamo puede renovarse."
    elif categoria == "Devolución":
        prioridad = "Media"
        accion = "Registrar y verificar la devolución del material."
    elif categoria == "Préstamo":
        prioridad = "Media"
        accion = "Revisar la solicitud y la disponibilidad del libro."
    elif categoria == "Consulta":
        prioridad = "Baja"
        accion = "Atender la consulta de información de biblioteca."
    else:
        prioridad = "Por evaluar"
        accion = "Revisar manualmente la solicitud para identificar su tipo."

    asunto = f"[{prioridad}] Solicitud de biblioteca: {categoria}"

    cuerpo = (
        "Buen día, equipo de soporte de biblioteca:\n\n"
        "Se ha recibido una nueva solicitud mediante el agente inteligente.\n\n"
        f"Mensaje del usuario: {mensaje}\n"
        f"Categoría identificada: {categoria}\n"
        f"Prioridad asignada: {prioridad}\n"
        f"Acción recomendada: {accion}\n\n"
        "Favor de revisar y dar seguimiento a la solicitud.\n\n"
        "Saludos,\n"
        "Agente inteligente de biblioteca"
    )

    return {
        "destinatario": destinatario,
        "asunto": asunto,
        "cuerpo": cuerpo
    }


def agente_biblioteca(mensaje):
    categoria = clasificar_solicitud(mensaje)
    correo = generar_correo_soporte(mensaje, categoria)

    resultado = {
        "mensaje_original": mensaje,
        "categoria": categoria,
        "correo_soporte": correo
    }

    return resultado


# Ejemplo de ejecución
mensaje = "No encuentro el libro que me prestaron"

resultado = agente_biblioteca(mensaje)

# Mostrar exclusivamente JSON válido
print(json.dumps(resultado, ensure_ascii=False, indent=4))
```

**EJEMPLO 1**

En este ejemplo, el agente inteligente recibe un mensaje, identifica el tipo de incidente, organiza la información en un objeto JSON y genera un correo dirigido al equipo de soporte.

```python
import json

# 1. Clasificar la solicitud
def clasificar_solicitud(mensaje):
    texto = mensaje.lower()

    if "no enciende" in texto or "pantalla" in texto:
        return "Hardware"

    elif "internet" in texto or "wifi" in texto:
        return "Conectividad"

    elif "contraseña" in texto or "contrasena" in texto:
        return "Acceso a cuenta"

    else:
        return "Solicitud no identificada"


# 2. Generar el correo para soporte
def crear_correo_soporte(mensaje, categoria):

    destinatario = "soporte@institucion.edu"

    if categoria == "Hardware":
        prioridad = "Alta"
        accion = "Revisar el equipo de cómputo."

    elif categoria == "Conectividad":
        prioridad = "Media"
        accion = "Verificar la conexión a internet."

    elif categoria == "Acceso a cuenta":
        prioridad = "Media"
        accion = "Orientar al usuario para recuperar su acceso."

    else:
        prioridad = "Por evaluar"
        accion = "Revisar manualmente la solicitud."

    asunto = f"[{prioridad}] Incidente de soporte: {categoria}"

    cuerpo = (
        "Buen día, equipo de soporte:\n\n"
        "Se ha recibido un nuevo incidente mediante el agente inteligente.\n\n"
        f"Mensaje recibido: {mensaje}\n"
        f"Categoría: {categoria}\n"
        f"Prioridad: {prioridad}\n"
        f"Acción recomendada: {accion}\n\n"
        "Favor de revisar y dar seguimiento al reporte.\n\n"
        "Saludos,\n"
        "Agente inteligente de soporte"
    )

    return {
        "destinatario": destinatario,
        "asunto": asunto,
        "cuerpo": cuerpo
    }


# 3. Integrar el agente inteligente
def agente_soporte(mensaje):

    categoria = clasificar_solicitud(mensaje)

    correo = crear_correo_soporte(
        mensaje,
        categoria
    )

    resultado = {
        "mensaje_original": mensaje,
        "categoria": categoria,
        "correo_soporte": correo
    }

    return resultado


# 4. Ejecutar el ejemplo
mensaje = "Mi computadora no enciende"

resultado = agente_soporte(mensaje)

# Mostrar exclusivamente el objeto JSON
print(json.dumps(resultado, ensure_ascii=False, indent=4))
```
